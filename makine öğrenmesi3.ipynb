```python
kenar_bir = int(input("birinci kenarı girin:"))
kenar_iki = int(input("ikinci kenarı girin:"))
kenar_uc = int(input("üçüncü kenarı girin:"))

try:
  if kenar_bir + kenar_iki > kenar_uc:
    if kenar_iki + kenar_uc > kenar_bir:
      if kenar_bir + kenar_uc > kenar_iki:
          print("üçgen oluşturulur")

    else:
      print("üçgen oluşturulamaz")

except ValueError:
  print("lütfen sayı girin")
```

```python
kullanici = "ilayda"
sifre = "0000"

kullanici_adi = input("kullanıcı adı girin:")
sifre_gir = input("şifre girin:")

if kullanici_adi == kullanici and sifre_gir == sifre:
  print("giriş başarılı")

elif kullanici_adi == kullanici and sifre_gir != sifre:
  print("şifre yanlış")

elif kullanici_adi != kullanici and sifre_gir == sifre:
  print("kullanıcı geçersiz")

else:
  print("hatalı giriş!")
```

```python
meyveler = ["elma" , "armut" , "muz"]


meyve_sayisi = 0
for meyve in meyveler :
  meyve_sayisi += 1
  print(f"{meyve_sayisi}.meyve {meyve}")
  print(f"{meyve_sayisi}.meyve {meyveler}")

for meyve, meyveler in enumerate(meyveler , start =1):
  print(f"{meyve} .meyve = {meyveler}")
```
