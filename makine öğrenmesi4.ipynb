```python
def asal_mi(sayi):
  if sayi < 2:
    return False

  for i in range(2,sayi):
    if sayi % i == 0:
      return False
  return True

asal_filtre = list(filter(asal_mi , range(1,51)))
print(asal_filtre)
```

```python
def topla(a, b):
    return a + b

def cikart(a, b):
    return a - b

def carp(a, b):
    return a * b

def bolme(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "0'a bölünemez!"

ISLEMLER = {
    '1': ('+', topla),
    '2': ('-', cikart),
    '3': ('*', carp),
    '4': ('/', bolme)
}

while True:
    print('\n=== HESAP MAKİNESİ ===')
    print("1. Toplama\n2. Çıkarma\n3. Çarpma\n4. Bölme\n5. Çıkış")

    secim = input("İşlem seçin: ").strip()      # ← döngünün İÇİNDE

    if secim == "5":
        print("Çıkış yapıldı.")
        break

    if secim in ISLEMLER:
        a = float(input("Birinci sayı: "))     # ← döngünün İÇİNDE, ayrı değişken
        b = float(input("İkinci sayı: "))      # ← döngünün İÇİNDE, ayrı değişken
        sembol, fonksiyon = ISLEMLER[secim]
        print(f"{a} {sembol} {b} = {fonksiyon(a, b)}")
    else:
        print("Geçersiz seçim!")
```
